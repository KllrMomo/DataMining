Pregunta D

```python
df['Size'].replace('Varies with device', np.nan, inplace=True)
```

¿Por qué es mejor convertirlo a NaN en esta etapa en lugar de ponerle un tamaño falso como "0 MB"?

Porque Varies with device no significa que pese 0 MB, si no que el tamaño depende del dispositivo, por lo que poner 0 MB podria afectar los calculos y el modelo.
