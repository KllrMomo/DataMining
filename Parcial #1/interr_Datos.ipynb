Pregunta B

```python
female_count = df.groupby('Sex').size()['female']
male_count = df.groupby('Sex').size()['male']

female_survived = df[(df['Sex'] == 'female') & (df['Survived'] == 1)].shape[0]
male_survived = df[(df['Sex'] == 'male') & (df['Survived'] == 1)].shape[0]

print("La cantidad de mujeres es de:", female_count, ", las que sobrevivieron es de:", female_survived)
print("La cantidad de hombres es de:", male_count, ", los que sobrevivieron es de:", male_survived)
```

¿Qué porcentaje exacto de mujeres sobrevivió en contraste con los hombres?

```python
porcentaje_mujeres = (female_survived / female_count) * 100
porcentaje_hombres = (male_survived / male_count) * 100

print("El porcentaje de mujeres que sobrevivieron es: ", porcentaje_mujeres, "%")
print("El porcentaje de hombres que sobrevivieron es: ", porcentaje_hombres, "%")
```

Hay una diferencia de 55.3130071% entre el porcentaje de hombres y mujeres

Pregunta C

```python
primer_cuartil = df['Fare'].quantile(0.25)
tercer_cuartil = df['Fare'].quantile(0.75)

IQR = tercer_cuartil - primer_cuartil

limite_superior = tercer_cuartil + (1.5 * IQR)


print("Primer cuartil: ", primer_cuartil)
print("Tercer cuartil: ", tercer_cuartil)
print("El límite superior del rango intercuartílico es: ", limite_superior)
```

```python
above_limit_fare = df[df['Fare'] > limite_superior]

print("Pasajeron que pagaron mas de lo normal: ", len(above_limit_fare))

print(above_limit_fare['Pclass'].value_counts())
```

La cantidad de pasajeron que pagaron mas de lo normal fueron los de primera clase, lo que tiene sentido ya que aun los tickets de primera clase son los que tienen un costo mas caro

Pregunta D

```python
mean_fare = df['Fare'].mean()
median_fare = df['Fare'].median()

print("El promedio de la tarifa es: ", mean_fare)
print("La mediana de la tarifa es: ", median_fare)
```

¿Qué significa que la media sea tan superior a la mediana?

Nos dice que en promedio la mayoria de los pasajeros pagaron menos, pero hay una cantidad de pasajeros que pagaron mas.

Si en el futuro utilizamos un algoritmo basado en distancias euclidianas (como K-Nearest Neighbors) sin escalar previamente esta variable, ¿cómo afectará esta asimetría al aprendizaje del modelo?

Podria tener un  efecto negativo, ya que podria tener una sensibilidad a las tarifas elevadas lo que puede generar una desproporcion.
