```python
# 26 - Ago
def female_fare():
    female = df[df['Sex'] == 'female']
    return female['Fare'].mean()

print(f"El promedio de las mujeres es: {female_fare()}")

def male_fare():
    male = df[df['Sex'] == 'male']
    return male['Fare'].mean()

print(f"El promedio de los hombres es: {male_fare()}")
```

```python
primer_cuartil = df['Fare'].quantile(0.25)
tercer_cuartil = df['Fare'].quantile(0.75)

IQR = tercer_cuartil - primer_cuartil
minimun = primer_cuartil - 1.5*IQR
maximum = tercer_cuartil + 1.5*IQR


print("Primer cuartil: ", primer_cuartil)
print("Tercer cuartil: ", tercer_cuartil)
print("IQR: ", IQR)
print("Mínimo: ", minimun)
print("Máximo: ", maximum)
```
