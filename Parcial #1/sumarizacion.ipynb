```python
import pandas as pd
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

----

Asimetría (Skewness)

```python
age = df['Age'].skew()
print("Asimetria de Age: ", age)
```

¿Qué significa físicamente sobre la edad de los pasajeros en el barco?

Hubo una simetria positiva, lo que nos dice que hubo mas pasajeros jovenes comprado con las personas mayores.

----

Curtosis (Kurtosis)

```python
fare = df['Fare'].kurt()
print("Kurtosis de Fare: ", fare)
```

¿Qué significa este número sobre el "grosor" de las colas de la distribución?

Es una dostribucion leprocurtica, lo que nos dice que es una distribucion muy puntiaguda y con colas pesadas en comparacion con una distribucion normal.

¿Confirma esto la existencia de los outliers que encontramos hoy en clase?

Si, ya que la curtosis es una medida sensible a la presencia de outliers.

----

Tablas Cruzadas (Cross-Tabulation)

```python
tabla = pd.crosstab(
    df['Pclass'], 
    df['Survived'], 
    margins=True, 
    margins_name="Total"
)

print("Frecuencias Absolutas:")
print(tabla)
```

¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

Se puede observar que hay una desigualdad economica entre la primera y la tercera clase, ya que por como va disminuyeno la clase de pasajeros el numero de fallecimientos incrementa.
