```python
import random 
random_list = [random.random() for _ in range(5_000_000)]
print("Random list length:", len(random_list))
```

```python
def SD(random_list, mean):
    return sum ((value - mean)**2 for value in random_list) / len(random_list)
print ("Standard Deviation:", SD(random_list, sum(random_list) / len(random_list)))
```

```python
#Hecho por profe
def mean_custom(values):
    accumulate = 0
    for val in values:
        accumulate += val
    return accumulate / len(values)

def st_deveation(values):
    sd = 0
    mean_val = mean_custom(values)
    for val in values:
        sd += (val - mean_val) ** 2

    sd = sd / len(values)

    return sd ** 0.5
```

```python
print(f"Std: (st_deveation): {st_deveation(random_list)}")
```

```python
import numpy as np

np_array = np.array(random_list)
print(f'Std: {np_array.std()}')
```
