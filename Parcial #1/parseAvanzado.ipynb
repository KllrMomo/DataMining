```python
import pandas as pd
import numpy as np

url = 'playstore_limpio.csv'
df = pd.read_csv(url)
```

----

### Reto 1: La Trampa de las Unidades de Medida

```python
def convertir_size(size):
    if pd.isna(size) or size == "Varies with device":
        return np.nan
    elif isinstance(size, (int, float)):
        return float(size)
    elif size.endswith("M"):
        return float(size[:-1])
    elif size.endswith("k"):
        return float(size[:-1]) / 1024
    return np.nan

df['Size'] = df['Size'].apply(convertir_size)

promedio = df['Size'].mean()

print("Promedio de Size:", promedio)
```

### Reto 2: El Tipo de Dato Cronológico

```python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])

df['Year_Updated'] = df['Last Updated'].dt.year

df["Year_Updated"].value_counts()
```
