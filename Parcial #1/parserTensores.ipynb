```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

----

Pregunta A

```python
duplicados = df.duplicated(subset=['App'])
print("Número de filas duplicadas: ", duplicados.sum())
```

```python
df.drop_duplicates(subset=['App'], inplace=True)
```

¿Cuántos registros duplicados purgaron?

Se encontrar y eliminaron 1181 registros duplicados.

----

Pregunta B

```python
df[df['Installs'] == 'Free']
```

```python
df = df[df['Installs'].ne('Free')].copy()

df['Installs'] = (
    df['Installs']
    .astype(str)
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(int)
)
```

```python
df['Installs'].mean()
```

¿Cuál es el promedio real de descargas?

El promedio de instalacion es aprox 7.7 millones de descargas por aplicacion.

Pregunta C

```python
df['Price'] = (
    df['Price']
    .str.replace('$', '', regex=False)
    .astype(float)
)
```
